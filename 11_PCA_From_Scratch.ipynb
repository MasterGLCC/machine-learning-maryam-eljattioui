```python
import numpy as np
import matplotlib.pyplot as plt
# X = [Caractéristique 1, Caractéristique 2]
# On simule des données qui suivent une ligne diagonale
X_train = np.array([
    [2.5, 2.4], [0.5, 0.7], [2.2, 2.9], [1.9, 2.2], [3.1, 3.0], 
    [2.3, 2.7], [2.0, 1.6], [1.0, 1.1], [1.5, 1.6], [1.1, 0.9]
])
class MyPCA:
    """
    Implémentation de l'Analyse en Composantes Principales (PCA).
    Objectif : Réduire la dimensionnalité tout en conservant le maximum de variance.
    """
    def __init__(self, n_components):
        """
        :param n_components: Nombre de dimensions souhaitées après réduction.
        """
        self.n_components = n_components
        self.components = None # Vecteurs propres (axes principaux)
        self.mean = None       # Moyenne (pour la normalisation)

    def fit(self, X):
        """ Étape d'apprentissage : Calcul des nouveaux axes de projection. """
        
        # 1. CENTRER LES DONNÉES (Soustraction de la moyenne)
        # C'est crucial pour que la matrice de covariance soit correcte.
        self.mean = np.mean(X, axis=0)
        X_centered = X - self.mean

        # 2. CALCULER LA MATRICE DE COVARIANCE
        # Elle montre comment les variables varient ensemble.
        # Formule : 1/(n-1) * X_transpose * X
        covariance_matrix = np.cov(X_centered.T)

        # 3. CALCULER LES VALEURS ET VECTEURS PROPRES (Eigenvalues & Eigenvectors)
        # Les vecteurs propres définissent la direction des nouveaux axes.
        # Les valeurs propres définissent la 'force' (variance) de chaque axe.
        eigenvalues, eigenvectors = np.linalg.eig(covariance_matrix)

        # 4. TRIER LES VECTEURS PROPRES
        # On veut les axes qui capturent le PLUS de variance en premier.
        # On trie par ordre décroissant des valeurs propres.
        eigenvectors = eigenvectors.T
        idxs = np.argsort(eigenvalues)[::-1]
        
        eigenvalues = eigenvalues[idxs]
        eigenvectors = eigenvectors[idxs]

        # 5. SÉLECTIONNER LES K COMPOSANTES
        # On ne garde que les 'n_components' premiers vecteurs.
        self.components = eigenvectors[0 : self.n_components]

    def transform(self, X):
        """ Projection des données originales sur les nouveaux axes. """
        # On centre les données avec la moyenne apprise lors du fit()
        X_centered = X - self.mean
        
        # Projection (Produit scalaire entre X et la transposée des composantes)
        # On passe d'un espace de dimension N à un espace de dimension K.
        return np.dot(X_centered, self.components.T)

# =================================================================
# 3. EXÉCUTION ET ANALYSE
# =================================================================
if __name__ == "__main__":
    # On veut passer de 2 dimensions à 1 seule dimension
    pca = MyPCA(n_components=1)

    # Entraînement
    pca.fit(X_train)

    # Transformation
    X_reduced = pca.transform(X_train)

    print("--- RÉSULTATS PCA ---")
    print(f"Dimensions originales : {X_train.shape}")
    print(f"Nouvelles dimensions : {X_reduced.shape}")
    print("\nDonnées projetées (1D) :")
    print(X_reduced)
```
